```python
import pyzx as zx
from pyzx.rewrite_rules.wrules_pcy import check_phase_copy_forward
```

```python
g = zx.Graph()

i = g.add_vertex(qubit=0.5, row=0)
z = g.add_vertex(ty=zx.VertexType.Z_BOX, qubit=0.5, row=1)
wi = g.add_vertex(ty=zx.VertexType.W_INPUT, qubit=0.5, row=2)
wo = g.add_vertex(ty=zx.VertexType.W_OUTPUT, qubit=0.5, row=3)
o0 = g.add_vertex(qubit=0, row=4)
o1 = g.add_vertex(qubit=1, row=4)

g.add_edge((i, z))
g.add_edge((z, wi))
g.add_edge((wi, wo))
g.add_edge((wo, o0))
g.add_edge((wo, o1))

print(f"PCY rule applicable to {wo} and {z} ? {check_phase_copy_forward(g, wo, z)}")

zx.draw(g, labels=True)
```

```python
g = zx.Graph()

i = g.add_vertex(qubit=0.5, row=0)
wi = g.add_vertex(ty=zx.VertexType.W_INPUT, qubit=0.5, row=1)
wo = g.add_vertex(ty=zx.VertexType.W_OUTPUT, qubit=0.5, row=2)

z0 = g.add_vertex(ty=zx.VertexType.Z_BOX, qubit=0, row=3)
o0 = g.add_vertex(qubit=0, row=4)

z1 = g.add_vertex(ty=zx.VertexType.Z_BOX, qubit=1, row=3)
o1 = g.add_vertex(qubit=1, row=4)

g.add_edge((i, wi))
g.add_edge((wi, wo))
g.add_edge((wo, z0))
g.add_edge((wo, z1))
g.add_edge((z0, o0))
g.add_edge((z1, o1))

zx.draw(g, labels=True)
```
